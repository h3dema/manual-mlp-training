This notebook is for testing the MLPRegressor on a regression problem.  
The dataset is a simple one with 2 features and 1 target variable.  
The goal is to predict the target variable based on the input features.  

> We use this to compare the performance of our MLP implementation with the sklearn MLPRegressor.

```python
%%capture
%pip install matplotlib scikit-learn pandas
```

```python
import pandas as pd
from sklearn.neural_network import MLPRegressor
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../main/regression.csv", header=None)
X = data.iloc[:, :-1]
y = data.iloc[:, -1]
```

```python
regr = MLPRegressor(
    random_state=1,
    max_iter=10000,
    tol=0.1,
    learning_rate='constant',
    learning_rate_init=0.001,
    hidden_layer_sizes=(16, 16))
_ = regr.fit(X, y)
```

```python
regr.score(X, y)
```

```python
y_pred = regr.predict(X)

plt.scatter(y, y_pred, color='blue', label='Predicted vs True Values', s=1, marker='+')
plt.xlabel('True Values')
plt.ylabel('Predicted Values')
plt.tight_layout()
_ = plt.show()
```
